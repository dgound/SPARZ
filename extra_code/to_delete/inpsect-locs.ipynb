```python
#%%
import matplotlib.pyplot as plt
import os

# Import your function from the local .py file
# This requires 'h5r_functions.py' to be in the same folder as your notebook.
from h5r_functions import h5r_to_df

print("Functions and libraries imported successfully.")
```

```python
#%%
def plot_h5r_comparison(filepath1, filepath2, x_col='fitResults_x0', y_col='fitResults_y0'):
    """
    Loads data from two h5r files using h5r_to_df and creates a scatter plot.
    """
    print(f"Loading and plotting data...")
    
    # Load the data from both files into pandas DataFrames
    df1 = h5r_to_df(filepath1)
    df2 = h5r_to_df(filepath2)

    # Create a plot
    fig, ax = plt.subplots(figsize=(10, 8))

    # Plot data from the first file
    ax.scatter(df1[x_col], df1[y_col], color='blue', label=os.path.basename(filepath1), s=1, alpha=0.1)

    # Plot data from the second file
    ax.scatter(df2[x_col], df2[y_col], color='red', label=os.path.basename(filepath2), s=1, alpha=0.1)

    # Add plot details
    ax.set_xlabel(x_col)
    ax.set_ylabel(y_col)
    ax.set_title(f'Comparison of {y_col} vs. {x_col}')
    ax.legend()
    ax.grid(True)
    ax.set_aspect('equal', adjustable='box')
    
    # Display the plot
    plt.show()

print("Plotting function is ready.")
```

```python
#%%
# --- 1. Define the paths to your two HDF5 files ---
file_path_1 = '/Volumes/Genetics/Wu_Lab-Vutara/Experiments/Laura/Colabs/compression/figure_1_data/synth_tubulin_data/raw/sequence-MT0.N1.HD-BP.h5r'
file_path_2 = '/Volumes/Genetics/Wu_Lab-Vutara/Experiments/Laura/Colabs/compression/figure_1_data/synth_tubulin_data/sparzffv1_level_0/new/sequence-MT0.N1.HD-BP.h5r'


# --- 2. Run the plotting function ---
# It will only run if both files are found at the specified paths.
if os.path.exists(file_path_1) and os.path.exists(file_path_2):
    plot_h5r_comparison(file_path_1, file_path_2)
else:
    print("ERROR: One or both file paths are incorrect. Please update the paths in this cell.")
    print(f"Checked for '{file_path_1}' (Exists: {os.path.exists(file_path_1)})")
    print(f"Checked for '{file_path_2}' (Exists: {os.path.exists(file_path_2)})")
```
