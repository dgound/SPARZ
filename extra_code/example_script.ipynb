```python
import os
import re

# The folder where your decompressed files are
folder_path = "X:\Wu_Lab-Vutara\Experiments\Laura\SPARZ\Figure3_redo\sparz_3\Location-05\Raw Images"

# Loop through every file in the folder
for filename in os.listdir(folder_path):
    old_file_path = os.path.join(folder_path, filename)
    
    # Skip folders, only look at files
    if not os.path.isfile(old_file_path):
        continue

    # Search for "img" followed by exactly 6 numbers (e.g., "img000000")
    match = re.search(r'(img\d{6})', filename)
    
    if match:
        clean_name = match.group(1)           # This grabs the "img000000" part
        ext = os.path.splitext(filename)[1]   # This grabs the ".dat" or ".tiff" part
        
        new_filename = f"{clean_name}{ext}"
        new_file_path = os.path.join(folder_path, new_filename)
        
        # Safety Check 1: If it's already named correctly, skip it
        if filename == new_filename:
            continue
            
        # Safety Check 2: Don't overwrite files if one already exists
        if os.path.exists(new_file_path):
            print(f"⚠️ Warning: {new_filename} already exists! Skipping {filename}.")
            continue
            
        # Actually rename the file
        os.rename(old_file_path, new_file_path)
        print(f"✅ Renamed: {filename}  -->  {new_filename}")

print("\nCleanup complete!")
```
