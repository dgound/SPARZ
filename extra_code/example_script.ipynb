```python
from SPARZ import UNSPARZ
```

```python
import os
import glob

num_files = 100
num_workers = 6
num_dask_workers = 3
output_format = 'dat'  
chunk_size = 10

# 1. Define the output directory
outpt = "X:/Wu_Lab-Vutara/Experiments/Laura/SPARZ/Figure3_redo/sparz_3_k21_th45/decompressed/Raw Images"

# 2. Safely make the directory
os.makedirs(outpt, exist_ok=True)

for i in range(num_files):
    _num = f"{i:06d}"
    
    # 1. Construct file paths with the current image number
    sparse_bp1 = f'X:/Wu_Lab-Vutara/Experiments/Laura/SPARZ/Figure3_redo/sparz_3_k21_th45/compressed/img{_num}*.npz'
    path_enc_bp1 = f'X:/Wu_Lab-Vutara/Experiments/Laura/SPARZ/Figure3_redo/sparz_3_k21_th45/compressed/img{_num}*.mp4'
    
    # 3. THE SAFETY CHECK (Prevents crashes if file is missing)
    if len(glob.glob(sparse_bp1)) == 0 or len(glob.glob(path_enc_bp1)) == 0:
        print(f"Skipping image {_num} (File not found)")
        continue

    print(f'\n--- Processing image {_num} ---')
    print('sparse_bp1:', sparse_bp1)
    print('path_enc_bp1:', path_enc_bp1)
    
    u = UNSPARZ(
        path_sparse_bp1=sparse_bp1,
        path_encoded_bp1=path_enc_bp1,
        stem=f'img{_num}',             
        output_path=outpt,
        path_sparse_bp2=None,
        path_encoded_bp2=None,
        use_roi=True,
        chunk_size=chunk_size,         
        num_workers=num_workers,       
        output_format=output_format,   
        num_dask_workers=num_dask_workers 
    )

    u.run()
```

```python
import os
import re

# The folder where your decompressed files are
folder_path = "X:\Wu_Lab-Vutara\Experiments\Laura\SPARZ\Figure3_redo\sparz_3_k21_th45\Location-05\Raw Images"
mkdir_cmd = f'mkdir "{folder_path}"'
os.system(mkdir_cmd)


# Loop through every file in the folder
for filename in os.listdir(folder_path):
    old_file_path = os.path.join(folder_path, filename)
    
    # Skip folders, only look at files
    if not os.path.isfile(old_file_path):
        continue

    # Search for "img" followed by exactly 6 numbers (e.g., "img000000")
    match = re.search(r'(img\d{6})', filename)
    
    if match:
        clean_name = match.group(1)           # This grabs the "img000000" part
        ext = os.path.splitext(filename)[1]   # This grabs the ".dat" or ".tiff" part
        
        new_filename = f"{clean_name}{ext}"
        new_file_path = os.path.join(folder_path, new_filename)
        
        # Safety Check 1: If it's already named correctly, skip it
        if filename == new_filename:
            continue
            
        # Safety Check 2: Don't overwrite files if one already exists
        if os.path.exists(new_file_path):
            print(f"⚠️ Warning: {new_filename} already exists! Skipping {filename}.")
            continue
            
        # Actually rename the file
        os.rename(old_file_path, new_file_path)
        print(f"✅ Renamed: {filename}  -->  {new_filename}")

print("\nCleanup complete!")
```
